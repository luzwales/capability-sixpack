## Notes

### Missing Values
Some sheets use `*` to indicate missing data:
- Pareto (Duration column)
- Student (Math grade column)

### Wide vs Long Format
Several sheets store groups in separate columns (wide format):
- Moisture (4 machine columns)
- Fertilizer (2 fertilizer columns)
- Productivity (3 shift columns)

Use `pd.melt()` to convert to long format for analysis.

### Metadata Rows
Each sheet has metadata in the first rows. The `skiprows` parameter skips to the actual data header row.
