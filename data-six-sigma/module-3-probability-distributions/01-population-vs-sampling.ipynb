# Population vs Sampling

**Module 3: Basic Statistics** | Lean Six Sigma Data Analytics

---

## The Problem

You lead a project to improve customer satisfaction. You want to know whether your customers are satisfied with the company's product.

Your organization has **40,000 customers**. How would you investigate if they're satisfied?

You could ask every single customer—but this is **impractical**. It's simply too much work.

**Solution:** Select a small group (say, 100 customers) to ask. This small group is called a **sample**.

---

## Key Definitions

| Term | Definition | Example |
|------|------------|----------|
| **Population** | All units of interest | All 40,000 customers |
| **Sample** | A subset of the population | 100 selected customers |

### Visual Representation

```
POPULATION (Unknown)          SAMPLE (Measured)
┌─────────────────────┐       ┌─────────────────┐
│  ?  ?  ?  ?  ?  ?   │       │  X₁  X₂  X₃    │
│  ?  ?  ?  ?  ?  ?   │  ──►  │  X₄  X₅  ...   │
│  ?  ?  ?  ?  ?  ?   │       │  X₉₉  X₁₀₀     │
│  ...40,000 units... │       │  100 units      │
└─────────────────────┘       └─────────────────┘
```

- **Question marks (?):** Unknown values in the population
- **X values:** Actual measurements from our sample

## The Statistical Process

```
┌──────────────┐     ┌──────────────┐     ┌──────────────┐     ┌──────────────┐
│  POPULATION  │     │   SAMPLE     │     │   ANALYZE    │     │    CLAIM     │
│              │ ──► │              │ ──► │              │ ──► │              │
│  ? ? ? ? ?   │     │  X₁ X₂ X₃   │     │  Mean, StDev │     │  About the   │
│  All units   │     │  Select 100  │     │  Histogram   │     │  population  │
└──────────────┘     └──────────────┘     └──────────────┘     └──────────────┘
```

1. **Population:** All units of interest (unknown)
2. **Sample:** Select units randomly for a representative sample
3. **Analyze:** Collect and analyze data from the sample
4. **Claim:** Use analysis to make claims about the entire population

---

## Example 1: Customer Satisfaction

| | Value |
|---|---|
| Population | 40,000 customers |
| Sample | 100 customers |
| Question | "Are you satisfied with our product?" |
| Result | 75 out of 100 said yes (75%) |

### The Key Question

Does 75% in our sample mean 75% of all 40,000 customers are satisfied?

**This is what statistics is all about:**
- Making claims about a population based on a sample
- Understanding how precise these claims are
- Factors like **sample size** affect precision

## Example 2: Coffee Batches

We want to measure caffeine percentage in coffee batches.

**What is the population?**

All batches produced:
- This morning
- This afternoon
- During the night shift
- Yesterday
- Tomorrow
- All future batches

**Population:** All batches of coffee produced, and to be produced

**Sample:** The specific batches we select to study (e.g., 40 batches)

We analyze the sample (histogram, mean, standard deviation), but we want to draw conclusions about **all** coffee batches—the entire population.

---

## Sample → Population

| From Sample | We Estimate | For Population |
|-------------|-------------|----------------|
| Sample distribution | → | Population distribution |
| Sample statistics | → | Population parameters |

### Important Notes

- **Representative sample:** Units must be selected **randomly** to be representative
- **Sample size matters:** Larger samples give more precise estimates
- We never know the true population values—we estimate them from samples

---

## Key Takeaways

- **Population:** All units of interest (often too large to measure completely)
- **Sample:** A subset selected from the population
- **Purpose:** Use sample data to make claims about the population
- **Random selection:** Essential for a representative sample
- **Sample size:** Affects the precision of our claims
- **Statistics:** The science of making claims about populations based on samples
