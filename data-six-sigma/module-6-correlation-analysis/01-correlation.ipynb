## What is Correlation?

Do you ever use the word correlation? Probably, you have. Now, let's explain what correlation means in a statistical sense and what it can be used for.

### Definition

> **Correlation** is an index that expresses how strong a relationship between two numerical variables is.

This index is always between **-1** and **+1**.

| Correlation | Meaning |
|-------------|--------|
| +1 | Perfect positive relationship |
| 0 | No linear relationship |
| -1 | Perfect negative relationship |

## Positive and Negative Correlation

### Positive Correlation

If you are studying two variables A and B, and they show a **positive correlation**, this means:

> Higher values of one variable generally correspond to **higher values** of the other variable.

### Negative Correlation

A **negative correlation** means the variables are correlated in the opposite direction:

> Higher values of one variable generally correspond to **lower values** of the other variable.

## Examples: Guessing the Correlation

Let's look at some examples. Can you guess the correlation in each graph?

```python
# Example 1: Strong positive correlation (r = 0.93)
n = 50
x1 = np.random.randn(n)
y1 = x1 * 0.95 + np.random.randn(n) * 0.3

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x1, y1, alpha=0.7, s=60)

# Add trend line
z = np.polyfit(x1, y1, 1)
p = np.poly1d(z)
ax.plot(np.sort(x1), p(np.sort(x1)), color=ms.RED, linestyle='--', alpha=0.8)

ax.set_xlabel('Variable A')
ax.set_ylabel('Variable B')
ax.set_title('Example 1: What is the correlation?')
plt.tight_layout()
plt.show()

corr1 = np.corrcoef(x1, y1)[0, 1]
print(f'Correlation = {corr1:.2f}')
print()
print('This is a STRONG POSITIVE correlation.')
print('  - The line is upward sloping (positive)')
print('  - The dots lie very close to the line (strong, close to 1)')
```

```python
# Example 2: Weak positive correlation (r = 0.31)
x2 = np.random.randn(n)
y2 = x2 * 0.3 + np.random.randn(n) * 0.9

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x2, y2, alpha=0.7, s=60)

# Add trend line
z = np.polyfit(x2, y2, 1)
p = np.poly1d(z)
ax.plot(np.sort(x2), p(np.sort(x2)), color=ms.RED, linestyle='--', alpha=0.8)

ax.set_xlabel('Variable A')
ax.set_ylabel('Variable B')
ax.set_title('Example 2: What is the correlation?')
plt.tight_layout()
plt.show()

corr2 = np.corrcoef(x2, y2)[0, 1]
print(f'Correlation = {corr2:.2f}')
print()
print('This is a WEAK POSITIVE correlation.')
print('  - You need a bit more imagination to see a straight line')
print('  - It is upward sloping (positive)')
print('  - But the dots are farther away from the line (weak)')
```

```python
# Example 3: No correlation (r ≈ 0)
x3 = np.random.randn(n)
y3 = np.random.randn(n)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x3, y3, alpha=0.7, s=60)

ax.set_xlabel('Variable A')
ax.set_ylabel('Variable B')
ax.set_title('Example 3: What is the correlation?')
plt.tight_layout()
plt.show()

corr3 = np.corrcoef(x3, y3)[0, 1]
print(f'Correlation = {corr3:.2f}')
print()
print('There is NO CORRELATION.')
print('  - The observations are just scattered around')
print('  - There is no relationship')
```

```python
# Example 4: Curved (non-linear) relationship
x4 = np.linspace(-3, 3, n)
y4 = x4**2 + np.random.randn(n) * 0.5

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x4, y4, alpha=0.7, s=60)

ax.set_xlabel('Variable A')
ax.set_ylabel('Variable B')
ax.set_title('Example 4: What is the correlation?')
plt.tight_layout()
plt.show()

corr4 = np.corrcoef(x4, y4)[0, 1]
print(f'Correlation = {corr4:.2f}')
print()
print('Can you see a relationship here?')
print()
print('Yes - but it is a CURVED relationship (non-linear).')
print()
print('We say there is NO CORRELATION because correlation is a LINEAR measure.')
print('The correlation coefficient does not capture non-linear relationships!')
```

## Interpretation: The Amsterdam Bike Example

In Amsterdam, many people take their bikes to work.

**Question:** Is the amount of rain related to the amount of people that travel to work by bike?

After some research, we find a **strong negative correlation**.

This means: The amount of people that travel by bike to work **decreases** as the amount of rain **increases**.

### Causation Warning

It is important to realize that we just **assumed** a cause and effect relationship.

We assumed that rain is the **cause** for people taking their bikes less.

However, **statistics alone will not help you with that choice**.

Be careful with recommendations like:

> *"More people should go by bike, because it will then rain less."*

In such a case, you have mistakenly assumed the **wrong cause and effect relationship** - you have basically turned it around!

## Rules of Thumb for Interpretation

How do you interpret a correlation coefficient?

| Correlation Coefficient | Interpretation |
|------------------------|----------------|
| 0 to 0.4 | No or weak correlation |
| 0.4 to 0.8 | Moderate correlation |
| 0.8 to 1.0 | Very strong relationship |
| Exactly 1 | Perfect correlation |

**Note:** These same thresholds apply to negative correlations (use absolute value).

In reality, it rarely occurs that two variables correlate perfectly (r = 1 or r = -1).

```python
# Visualize the rules of thumb
print('INTERPRETING CORRELATION COEFFICIENTS')
print('=' * 50)
print()
print('|  Correlation  |  Interpretation          |')
print('|---------------|--------------------------|')
print('|  0.0 - 0.4    |  No or weak correlation  |')
print('|  0.4 - 0.8    |  Moderate correlation    |')
print('|  0.8 - 1.0    |  Very strong correlation |')
print('|  1.0          |  Perfect correlation     |')
print()
print('(Same thresholds apply for negative correlations)')
```

## Summary

### Key Takeaways

1. **Correlation** is an index between -1 and +1 that measures the strength of a **linear** relationship between two numerical variables

2. **Positive correlation:** Higher values of one variable correspond to higher values of the other

3. **Negative correlation:** Higher values of one variable correspond to lower values of the other

4. **Correlation = 0** means no linear relationship (but there could be a non-linear relationship!)

5. **Rules of thumb:** 0-0.4 weak, 0.4-0.8 moderate, 0.8-1.0 very strong

### Final Warning

> **Always be careful not to mix up correlation with causality.**

Correlation tells you that two variables are related, but it does NOT tell you that one causes the other.
