## Example: Tea Bag Defects

Imagine you work at a factory that produces tea. You work for the packaging line where the tea is put into tea bags.

**Problem:** Sometimes the tea bags break during the packaging process and become defects. You wish to reduce the number of defect tea bags.

**Brainstorming:** You know that different tastes of tea are produced, and therefore the production line is stopped during the day for change-overs of type of tea.

**Question:** Could there be a relationship between the number of production stops and the number of defect tea bags?

You collect data on both variables.

```python
# Load tea bag data
def load_teabags():
    df = pd.read_excel(xlsx, sheet_name='Tea bags', skiprows=7)
    return df.dropna()

teabags = load_teabags()

print('TEA BAG DATA')
print('=' * 40)
print()
print('Variables:')
print('  Y (CTQ): Bags - Number of defective tea bags')
print('  X: Stops - Number of production stops')
print()
print(teabags.head(10))
```

---

## Visualizing the Relationship: Scatter Plot

To study the relationship, we first look at a graph of this data.

- **Y variable:** Defect bags (called "Bags" in the dataset)
- **X variable:** Stops

Because they are both **numerical**, we can make a **scatter plot** to visualize the relationship.

```python
# Scatter plot
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(teabags['Stops'], teabags['Bags'], s=80, alpha=0.7)

ax.set_xlabel('Stops (Number of production stops)')
ax.set_ylabel('Bags (Number of defective tea bags)')
ax.set_title('Scatter Plot: Defective Tea Bags vs Production Stops')

plt.tight_layout()
plt.show()

print('Do you see a relationship between Stops and Defects?')
print()
print('On first sight, it looks like the number of broken tea bags')
print('INCREASES as the number of production stops increases as well.')
```

---

## Finding the Best Fitting Line

We can visualize this relationship by drawing a line through the data points.

**But which line is the best line?**

```python
# Show different possible lines
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

x = teabags['Stops'].values
y = teabags['Bags'].values

# Different candidate lines
lines = [
    (1.5, 10, 'Is it this one?'),
    (2.5, 5, 'Or this one?'),
    (1.0, 15, 'Or maybe this one?')
]

for ax, (slope, intercept, title) in zip(axes, lines):
    ax.scatter(x, y, s=80, alpha=0.7)
    x_line = np.array([x.min(), x.max()])
    y_line = intercept + slope * x_line
    ax.plot(x_line, y_line, 'r-', linewidth=2)
    ax.set_xlabel('Stops')
    ax.set_ylabel('Bags')
    ax.set_title(title)

plt.tight_layout()
plt.show()

print('Regression analysis is about finding the BEST FITTING linear line for your data.')
```

---

## The Regression Equation

A line can be described mathematically by the formula:

$$y = a + bX$$

Where:
- **a** = intercept (where the line crosses the Y-axis)
- **b** = slope (how much Y changes when X increases by 1)

**Regression analysis means finding the a and b that give you the best fitting line.**

---

## The Four Steps of Regression Analysis

Regression analysis consists of **four steps**:

| Step | Description |
|------|-------------|
| **Step 1** | Make a **fitted line plot** to visualize your data |
| **Step 2** | Perform the **main regression analysis** - check if relationship is statistically significant and how strong it is |
| **Step 3** | Perform a **residual analysis** - the reliability of the regression depends on this step |
| **Step 4** | *(Optional)* Construct a **prediction interval** - if you want to use regression for predictions |

How to perform these four steps will be discussed in the next videos.

## Summary

### What is Regression Analysis?

Regression analysis is a statistical method to test if two **numerical variables** are related by calculating the **best fitting line**.

### The Four Steps

1. **Fitted line plot** - Visualize the data
2. **Main regression analysis** - The actual calculations (significance and strength)
3. **Residual analysis** - Check reliability
4. **Prediction interval** - Calculate predictions (optional)
