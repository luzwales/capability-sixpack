## Example: Coffee Caffeine Content

**Context:** You produce coffee and want to sell it as decaf. If caffeine percentage is below 0.1%, the batch qualifies as decaf.

**Data:** 40 batches measured for caffeine percentage.

**Question:** How well does the production process perform? Are we producing decaf coffee?

```python
# Load Caffeine batches data
caffeine = pd.read_excel(xlsx, sheet_name='Caffeine_batches', skiprows=6, usecols=[0, 1])
caffeine.columns = ['Batch', 'Caffeine_Pct']
caffeine = caffeine.dropna()

print(f'Dataset: {caffeine.shape[0]} batches\n')
caffeine.head(10)
```

---

## Two Categories of Descriptive Statistics

| Category | Measures | What It Tells You |
|----------|----------|-------------------|
| **Location** | Mean, Median | Where the data is centered |
| **Dispersion** | Standard Deviation, Variance, Range, IQR | How spread out the data is |

```python
# Quick summary using pandas describe()
# Equivalent to Minitab: Stat > Basic Statistics > Display Descriptive Statistics
caffeine['Caffeine_Pct'].describe()
```

```python
# Full descriptive statistics matching Minitab output
data = caffeine['Caffeine_Pct']

descriptive_stats = {
    'N': data.count(),
    'N* (missing)': data.isna().sum(),
    'Mean': data.mean(),
    'StDev': data.std(),
    'Variance': data.var(),
    'Minimum': data.min(),
    'Q1 (25th pctl)': data.quantile(0.25),
    'Median': data.median(),
    'Q3 (75th pctl)': data.quantile(0.75),
    'Maximum': data.max(),
    'Range': data.max() - data.min(),
    'IQR': data.quantile(0.75) - data.quantile(0.25)
}

stats_df = pd.DataFrame.from_dict(descriptive_stats, orient='index', columns=['Value'])
stats_df.round(6)
```

---

## Location Measures

### Mean (Average)

**Formula:** Sum of all measurements ÷ N

The mean indicates where your data is centered on the scale.

### Median

The value that separates the smallest 50% from the largest 50% when data is ordered.

| Measure | Advantage | Disadvantage |
|---------|-----------|--------------|
| Mean | Uses all data points; more precise | Affected by outliers |
| Median | Not affected by outliers | Only uses middle value; less precise |

```python
print('Location Measures:')
print(f'  Mean:   {data.mean():.4f}%')
print(f'  Median: {data.median():.4f}%')
print()
print(f'Mean and median are close, suggesting symmetric distribution.')
```

---

## Dispersion Measures

### Understanding Variation

Two datasets can have the same mean but very different spreads:
- **Low dispersion:** Values cluster close to the mean
- **High dispersion:** Values spread far from the mean

### Standard Deviation

**The Problem:** If we average the distances from each point to the mean, positives and negatives cancel out (≈ 0).

**The Solution:**
1. Square each distance (makes all values positive)
2. Average the squared distances = **Variance**
3. Take the square root = **Standard Deviation**

### Range vs IQR

| Measure | Formula | Robustness |
|---------|---------|------------|
| Range | Max − Min | Sensitive to outliers |
| IQR | Q3 − Q1 | Ignores extreme 25% on each end |

```python
print('Dispersion Measures:')
print(f'  Std Dev:  {data.std():.4f}%')
print(f'  Variance: {data.var():.6f}')
print(f'  Range:    {data.max() - data.min():.4f}%')
print(f'  IQR:      {data.quantile(0.75) - data.quantile(0.25):.4f}%')
print()
print(f'Interpretation: Observations are, on average, {data.std():.4f}% away from the mean.')
```

---

## Quartiles Explained

When data is ordered from small to large:

| Quartile | Definition |
|----------|------------|
| Q1 (25th percentile) | Separates lowest 25% from highest 75% |
| Median (50th percentile) | Separates lowest 50% from highest 50% |
| Q3 (75th percentile) | Separates lowest 75% from highest 25% |

```python
print('Quartiles:')
print(f'  Q1 (25%): {data.quantile(0.25):.4f}% - 25% of batches below this')
print(f'  Median:   {data.median():.4f}% - 50% of batches below this')
print(f'  Q3 (75%): {data.quantile(0.75):.4f}% - 75% of batches below this')
```

---

## Interpreting the Results

Now let's answer our business question: **Are we producing decaf coffee?**

```python
threshold = 0.1  # Decaf threshold

print('Business Interpretation')
print('=' * 50)
print(f'Decaf threshold: {threshold}%')
print()
print(f'Mean caffeine:    {data.mean():.4f}% - {"BELOW" if data.mean() < threshold else "ABOVE"} threshold')
print(f'Maximum caffeine: {data.max():.4f}% - {"BELOW" if data.max() < threshold else "ABOVE"} threshold')
print()

failing = (data > threshold).sum()
print(f'Batches exceeding threshold: {failing} out of {len(data)}')
print(f'Failure rate: {(data > threshold).mean() * 100:.1f}%')
print()
print('Conclusion:')
if data.mean() < threshold and failing == 0:
    print('  All batches qualify as decaf.')
elif data.mean() < threshold:
    print('  On average, coffee is decaf, but some batches fail.')
    print('  High dispersion is causing quality issues.')
else:
    print('  Average caffeine exceeds threshold - process needs improvement.')
```

## Python Quick Reference

| Statistic | Python Code |

---

## Key Takeaways

- **Descriptive statistics** summarize data into key figures before deeper analysis
- Only applicable to **numerical data**
- **Location measures:** Mean (uses all data, affected by outliers) and Median (robust)
- **Dispersion measures:** Std Dev/Variance (average distance from mean), Range/IQR (spread)
- **IQR** is more robust than range because it ignores extreme values
- Always examine both location AND dispersion to understand your data
