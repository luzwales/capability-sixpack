## The First Question

A useful start to any data analysis:

> **What type of variables am I studying? Are they numerical or categorical?**

This determines which visualizations and statistical techniques apply.

````python
# Setup
import pandas as pd
import numpy as np

DATA_FILE = '../data/da-lss.xlsx'
xlsx = pd.ExcelFile(DATA_FILE)
print('Data loaded successfully!')
````

---

## Two Main Categories of Data

| Category | Also Known As | Subdivisions |
|----------|---------------|--------------|
| **Numerical** | Quantitative | Discrete, Continuous |
| **Categorical** | Qualitative | Ordinal, Nominal, Binary |

### Numerical Data

**Data based on numbers that you can calculate with.**

| Type | Definition | Example |
|------|------------|---------|
| **Discrete** | Count data; whole numbers | Number of defects, iterations |
| **Continuous** | Any value on a scale | Temperature, time, weight |

### Categorical Data

**Data divided into groups.**

| Type | Definition | Example |
|------|------------|---------|
| **Ordinal** | Groups with meaningful order | Satisfaction (Low/Medium/High) |
| **Nominal** | Unordered categories | Reclaim type, Color, Department |
| **Binary** | Only two options | Yes/No, Pass/Fail |

---

## Example: Bank Reclaims Dataset

International money transfers sometimes have errors, leading clients to submit reclaims. Let's classify the variables in this dataset.

````python
# Load Reclaims data
reclaims = pd.read_excel(xlsx, sheet_name='Reclaims', skiprows=6, usecols=[0, 1, 2])
reclaims.columns = ['Reclaim_Type', 'Person', 'Processing_Time']
reclaims = reclaims.dropna()

print(f'Dataset: {reclaims.shape[0]} reclaims\n')
reclaims.head(10)
````

````python
# Classify each variable
print('Variable Classification:')
print('=' * 60)
print(f'Reclaim_Type: Categorical (Nominal)')
print(f'  - {reclaims["Reclaim_Type"].nunique()} unique categories')
print(f'  - No meaningful order between types')
print()
print(f'Person: Categorical (Nominal)')
print(f'  - {reclaims["Person"].nunique()} unique people')
print(f'  - Names have no inherent order')
print()
print(f'Processing_Time: Numerical (Continuous)')
print(f'  - Can be any value, not just integers')
print(f'  - Range: {reclaims["Processing_Time"].min():.1f} to {reclaims["Processing_Time"].max():.1f}')
````

---

## Why Does This Distinction Matter?

### Reason 1: Information Density

**Numerical data contains more information than categorical data.**

Consider tracking hospital patients who should leave within 6 days:

| Approach | Data Type | What You Record |
|----------|-----------|-----------------|
| Option A | Binary (Categorical) | Left within 6 days: Yes/No |
| Option B | Numerical | Duration: 4, 7, 5, 3, 8 days |

From numerical data, you can always derive the categorical version. The reverse is impossible.

> **Always collect data numerically when you have a choice.**

### Reason 2: Sample Size Requirements

| Data Type | Minimum Observations |
|----------|---------------------|
| Numerical | ~30 observations |
| Categorical | ~300+ observations |

### Reason 3: Determines Your Tools

| Variable Type | Appropriate Visualizations |
|---------------|---------------------------|
| **Numerical** | Histogram, Boxplot |
| **Categorical** | Pie chart, Bar chart, Pareto chart |

````python
# For categorical data: frequency tables
print('Frequency Table: Reclaim Types')
print('=' * 40)
freq = reclaims['Reclaim_Type'].value_counts()
freq_pct = (freq / freq.sum() * 100).round(1)

freq_table = pd.DataFrame({
    'Count': freq,
    'Percent': freq_pct
})
freq_table
````

````python
# For numerical data: descriptive statistics
print('Descriptive Statistics: Processing Time')
print('=' * 40)
reclaims['Processing_Time'].describe()
````

## Decision Tree

```
Is the variable based on numbers you can calculate with?
│
├── YES → NUMERICAL
│         │
│         ├── Can it only be whole numbers (counts)?
│         │   ├── YES → Discrete
│         │   └── NO → Continuous
│
└── NO → CATEGORICAL
          │
          ├── Only two options? → Binary
          ├── Meaningful order? → Ordinal
          └── No order? → Nominal
```

---

## Key Takeaways

- **First question in any analysis:** What type of variables am I studying?
- **Numerical** (quantitative): Numbers you calculate with → Discrete or Continuous
- **Categorical** (qualitative): Data in groups → Ordinal, Nominal, or Binary
- Numerical data is more information-rich; collect numerically when possible
- Variable type determines which charts and statistical tests to use
